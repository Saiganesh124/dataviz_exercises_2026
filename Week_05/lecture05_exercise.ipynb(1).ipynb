## Task 1 — Histogram: price by room type (overlapping distributions)

**What to build:** A histogram showing price distributions for **Entire home/apt vs Private room** (exclude Shared room — too few observations) overlaid on the same chart.

**Requirements:**
- Both room types on the same chart (use `color='room_type'`)
- `barmode='overlay'` with `opacity=0.6` so both distributions are visible
- A vertical line for the median of EACH room type, differently coloured
- Insight title comparing the two distributions

```python
import pandas as pd
import plotly.express as px

df.columns = df.columns.str.strip()

df['price'] = pd.to_numeric(
    df['price'].astype(str).str.replace(r'[$,£]', '', regex=True),
    errors='coerce'
)

df = df.dropna(subset=['price'])

p95 = df['price'].quantile(0.95)

df_cap = df[df['price'] <= p95].copy()

room_data = df_cap[
    df_cap['room_type'].isin(['Entire home/apt', 'Private room'])
].copy()

if room_data.empty:
    raise ValueError("No listings found for the selected room types.")

median_entire = room_data.loc[
    room_data['room_type'] == 'Entire home/apt', 'price'
].median()

median_private = room_data.loc[
    room_data['room_type'] == 'Private room', 'price'
].median()

fig = px.histogram(
    room_data,
    x='price',
    color='room_type',
    barmode='overlay',
    opacity=0.6,
    nbins=50,
    color_discrete_map={
        'Entire home/apt': '#636EFA',
        'Private room': '#EF553B'
    },
    title='Price Distribution: Entire Homes vs Private Rooms'
)

fig.add_vline(
    x=median_entire,
    line_color='blue',
    line_dash='dash'
)

fig.add_vline(
    x=median_private,
    line_color='red',
    line_dash='dash'
)

fig.add_annotation(
    x=median_entire,
    y=1,
    yref='paper',
    text=f'Entire home median: £{median_entire:.0f}',
    showarrow=False,
    yshift=15
)

fig.add_annotation(
    x=median_private,
    y=1,
    yref='paper',
    text=f'Private room median: £{median_private:.0f}',
    showarrow=False,
    yshift=35
)

fig.update_layout(
    xaxis_title='Price per Night (£)',
    yaxis_title='Number of Listings',
    legend_title='Room Type',
    template='plotly_white'
)

fig.show()
```

## Task 2 — Box plot: listing activity by borough

**What to build:** A **horizontal box plot** comparing listing activity (reviews per month) across London boroughs — reviews per month is a proxy for how frequently a listing is booked.

**Requirements:**
- Horizontal orientation (borough names are long)
- Sorted by median reviews per month (most active at top)
- Highlight the **two most active** boroughs in a different colour
- Outliers shown as individual points
- Insight title naming the two busiest boroughs

> 💡 Some listings have zero reviews — these are new or inactive listings. Filter them out with before plotting

```python
print(df.columns.tolist())
```

```python
import pandas as pd
import plotly.express as px

df.columns = df.columns.str.strip()

possible_columns = [
    'neighbourhood_cleansed',
    'neighbourhood',
    'neighbourhood_group_cleansed',
    'neighborhood',
    'borough'
]

location_column = next(
    (col for col in possible_columns if col in df.columns),
    None
)

if location_column is None:
    raise ValueError(
        "No neighbourhood column found. Check the output of print(df.columns.tolist())."
    )

df['reviews_per_month'] = pd.to_numeric(
    df['reviews_per_month'], errors='coerce'
)

activity_data = df[
    df['reviews_per_month'].notna() &
    (df['reviews_per_month'] > 0) &
    df[location_column].notna()
].copy()

borough_medians = (
    activity_data.groupby(location_column)['reviews_per_month']
    .median()
    .sort_values(ascending=False)
)

if len(borough_medians) < 2:
    raise ValueError("The dataset must contain at least two neighbourhoods.")

top_two = borough_medians.head(2).index.tolist()

borough_order = borough_medians.index.tolist()

activity_data['activity_group'] = activity_data[location_column].apply(
    lambda x: 'Top 2 busiest areas' if x in top_two else 'Other areas'
)

overall_median = activity_data['reviews_per_month'].median()

fig = px.box(
    activity_data,
    x='reviews_per_month',
    y=location_column,
    color='activity_group',
    orientation='h',
    points='outliers',
    category_orders={
        location_column: borough_order,
        'activity_group': ['Top 2 busiest areas', 'Other areas']
    },
    color_discrete_map={
        'Top 2 busiest areas': '#EF553B',
        'Other areas': '#636EFA'
    },
    title=f'Busiest Areas: {top_two[0]} and {top_two[1]}'
)

fig.add_vline(
    x=overall_median,
    line_dash='dash',
    line_color='black',
    annotation_text=f'Overall median: {overall_median:.2f}'
)

fig.update_layout(
    yaxis={'autorange': 'reversed'},
    xaxis_title='Reviews per Month',
    yaxis_title='London Neighbourhood',
    legend_title='Neighbourhood Activity',
    template='plotly_white'
)

fig.show()

print("Location column used:", location_column)
print("Top two busiest areas:", top_two)
```
